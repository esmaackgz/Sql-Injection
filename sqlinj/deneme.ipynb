```python
import sqlite3

# Veritabanı bağlantısı oluştur
conn = sqlite3.connect('users.db')
cursor = conn.cursor()

# Kullanıcı verilerini al
def get_user_data(username):
    cursor.execute("SELECT * FROM users WHERE username = ?", (username,))
    user_data = cursor.fetchone()
    return user_data

# Ana program
if __name__ == "__main__":
    # Kullanıcı adı al
    username = input("Kullanıcı adını girin: ")

    # Kullanıcı verilerini al
    user_data = get_user_data(username)

    if user_data:
        print("Kullanıcı verileri:")
        print("ID:", user_data[0])
        print("Kullanıcı adı:", user_data[1])
        print("E-posta:", user_data[2])
    else:
        print("Kullanıcı bulunamadı.")

# Bağlantıyı kapat
conn.close()
```

```python
import tkinter as tk
from tkinter import ttk

# Tkinter penceresini oluştur
root = tk.Tk()
root.title("Treeview'de Uzun Metin Gösterimi")

# Treeview için sütun tanımla
columns = ('Metin',)

# Treeview oluştur
tree = ttk.Treeview(root, columns=columns, show='headings')
tree.heading('Metin', text='Metin')

# Sütun genişliğini ayarla
tree.column('Metin', width=400)  # Metin sütununun genişliğini ayarla

# Örnek veriler ekle (uzun metin)
uzun_metin = "Bu çok uzun bir metindir ve Treeview'de tam olarak görüntülenmesi için sütun genişliği ayarlanmıştır."
tree.insert('', 'end', text=uzun_metin)

# Yatay kaydırma çubuğunu etkinleştir
yscroll = ttk.Scrollbar(root, orient='horizontal', command=tree.xview)
tree.configure(xscrollcommand=yscroll.set)
yscroll.pack(side='bottom', fill='x')

# Treeview'ı göster
tree.pack(expand=True, fill='both')

# Pencereyi göster
root.mainloop()
```
